# Download and load a GGUF with llama.cpp

This recreates the notebook pattern: download a GGUF from Hugging Face into a local `models/` directory, then load it with `llama_cpp.Llama`.

```python
%pip install -q -U llama-cpp-python huggingface_hub
```

```python
from huggingface_hub import hf_hub_download
from llama_cpp import Llama

REPO_ID = 'TheBloke/Llama-2-7B-Chat-GGUF'
FILENAME = 'llama-2-7b-chat.Q4_K_M.gguf'
MODEL_DIR = 'models'

model_path = hf_hub_download(
    repo_id=REPO_ID,
    filename=FILENAME,
    local_dir=MODEL_DIR,
)
print(model_path)
```

```python
llm = Llama(
    model_path=model_path,
    n_ctx=2048,
    n_gpu_layers=0,  # change to -1 for maximum GPU offload
    verbose=False,
)
```

```python
result = llm.create_chat_completion(
    messages=[{'role': 'user', 'content': 'What is GGUF?'}],
    max_tokens=100,
    temperature=0.2,
)
print(result['choices'][0]['message']['content'])
```
